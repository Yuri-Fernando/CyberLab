## 2. Implementar fail2ban

```python
print("\n🛡️ IMPLEMENTANDO FAIL2BAN\n")

fail2ban_config = """
[DEFAULT]
bantime = 3600
findtime = 600
maxretry = 3

[sshd]
enabled = true
port = ssh
filter = sshd
maxretry = 3

[ftp-aggressive]
enabled = true
port = ftp
maxretry = 3

[samba]
enabled = true
port = 445
maxretry = 3
"""

print("fail2ban Configuration:")
print("========================")
print(fail2ban_config)

print("\n📝 Para aplicar no Linux:")
print("  1. sudo apt-get install fail2ban")
print("  2. Editar /etc/fail2ban/jail.local")
print("  3. sudo systemctl restart fail2ban")
print("  4. sudo fail2ban-client status")
```

## 3. Políticas de Senha Forte

```python
print("\n🔐 POLÍTICA DE SENHAS FORTES\n")

password_policy = {
    "Min Length": "12 caracteres",
    "Complexidade": "Maiúsculas + minúsculas + números + símbolos",
    "Validade": "Mudar a cada 90 dias",
    "Histórico": "Não reutilizar últimas 5 senhas",
    "Account Lockout": "Bloquear após 3 tentativas falhadas",
    "Lockout Duration": "30 minutos",
    "MFA": "Obrigatório para admin"
}

for politica, valor in password_policy.items():
    print(f"  ✓ {politica}: {valor}")
```

## 4. Hardening de Serviços

```python
print("\n⚙️ HARDENING DE SERVIÇOS\n")

hardening_steps = {
    "SSH": [
        "Mudar porta padrão de 22 para 2222",
        "Desabilitar root login",
        "Usar autenticação por chave (não password)",
        "Desabilitar versionamento (Hide SSH version)"
    ],
    "FTP": [
        "Desabilitar anonymous login",
        "Usar SFTP (secure) ao invés de FTP",
        "Restringir acesso por IP",
        "Usar TLS/SSL"
    ],
    "SMB": [
        "Desabilitar SMB v1",
        "Usar SMB v3 com encriptação",
        "Desabilitar NetBIOS",
        "Restringir acesso"
    ],
    "HTTP": [
        "Forçar HTTPS",
        "Implementar rate limiting",
        "Remover banners de servidor",
        "Implementar WAF (Web App Firewall)"
    ]
}

for service, steps in hardening_steps.items():
    print(f"{service}:")
    for step in steps:
        print(f"  ✓ {step}")
    print()
```
