## 6. Verificar Ferramentas de Pentest

```python
def check_tool(tool_name, command):
    """Verifica se uma ferramenta está instalada."""
    try:
        result = subprocess.run(command, shell=True, capture_output=True, text=True, timeout=5)
        if result.returncode == 0:
            return True, "Instalado"
        else:
            return False, "Instalado mas com erro"
    except FileNotFoundError:
        return False, "Não encontrado"
    except subprocess.TimeoutExpired:
        return False, "Timeout"
    except Exception as e:
        return False, str(e)

tools = {
    "Nmap": "nmap --version",
    "Medusa": "medusa -h 2>&1 | head -1",
    "curl": "curl --version",
    "wget": "wget --version",
    "git": "git --version"
}

print("Verificando ferramentas de pentest:\n")

for tool, cmd in tools.items():
    installed, msg = check_tool(tool, cmd)
    status = "✅" if installed else "❌"
    print(f"{status} {tool:15} - {msg}")

print("\n⚠️ Em Linux/Kali, instale ferramentas faltantes com:")
print("   sudo apt-get install nmap medusa")
```

## 7. Criar Wordlists Básicas

```python
# Criar wordlist de usuários padrão
users = [
    "root",
    "admin",
    "administrator",
    "msfadmin",
    "user",
    "test",
    "ftpuser",
    "mysql",
    "postgres",
    "oracle"
]

passwords = [
    "password",
    "123456",
    "12345678",
    "admin",
    "root",
    "msfadmin",
    "user",
    "test",
    "password123",
    "admin123"
]

print("Criando wordlists...\n")

# FTP
ftp_users_file = wordlists_dir / "ftp_users.txt"
with open(ftp_users_file, "w") as f:
    f.write("\n".join(users))
print(f"✅ {ftp_users_file.name} ({len(users)} usuários)")

ftp_passwords_file = wordlists_dir / "ftp_passwords.txt"
with open(ftp_passwords_file, "w") as f:
    f.write("\n".join(passwords))
print(f"✅ {ftp_passwords_file.name} ({len(passwords)} senhas)")

# SMB
smb_users_file = wordlists_dir / "smb_users.txt"
with open(smb_users_file, "w") as f:
    f.write("\n".join(users))
print(f"✅ {smb_users_file.name} ({len(users)} usuários)")

smb_passwords_file = wordlists_dir / "common_passwords.txt"
with open(smb_passwords_file, "w") as f:
    f.write("\n".join(passwords))
print(f"✅ {smb_passwords_file.name} ({len(passwords)} senhas)")

print("\n✅ Wordlists criadas!")
```
