## 1. Setup

```python
import sys
import os
from pathlib import Path
from datetime import datetime
from dotenv import load_dotenv

# Setup
scripts_dir = Path("../scripts/python")
sys.path.insert(0, str(scripts_dir.absolute()))
load_dotenv(Path("../.env"))

TARGET_IP = os.getenv("TARGET_IP", "192.168.56.101")
WORDLISTS_DIR = Path("../wordlists")

print(f"✅ Setup concluído")
print(f"   Target: {TARGET_IP}")
print(f"   Wordlists: {WORDLISTS_DIR}")
```

## 2. OSINT - Open Source Intelligence

```python
print("🔍 OSINT - Pesquisa Pública sobre o Target\n")

osint_findings = {
    "target": TARGET_IP,
    "type": "Interno (192.168.56.0/24)",
    "purpose": "Metasploitable 2 - Sistema propositalmente vulnerável",
    "known_vulnerabilities": [
        "Versão antiga de FTP (vsftpd 2.0.1)",
        "Versão antiga de OpenSSH",
        "Samba com vulnerabilidades conhecidas",
        "DVWA em HTTP (sem HTTPS)",
        "MySQL acessível externamente"
    ],
    "likely_os": "Linux (Debian/Ubuntu antiga)",
    "likely_users": [
        "root",
        "msfadmin",
        "ftp",
        "mail",
        "www-data",
        "mysql",
        "postgres"
    ],
    "likely_passwords": [
        "msfadmin",  # Default Metasploitable
        "password",
        "admin",
        "root",
        "test",
        "12345678"
    ]
}

print("📊 Descobertas OSINT:\n")
print(f"Target: {osint_findings['target']}")
print(f"Tipo: {osint_findings['type']}")
print(f"Propósito: {osint_findings['purpose']}")

print(f"\nVulnerabilidades Conhecidas:")
for vuln in osint_findings['known_vulnerabilities']:
    print(f"  • {vuln}")

print(f"\nUsuários Prováveis:")
for user in osint_findings['likely_users']:
    print(f"  • {user}")

print(f"\nSenhas Prováveis:")
for pwd in osint_findings['likely_passwords']:
    print(f"  • {pwd}")
```

## 3. Geração de Wordlists Inteligentes

```python
print("📝 Gerando Wordlists Inteligentes\n")

# Tier 1: Padrão (mais provável)
tier1_users = [
    "root",
    "admin",
    "administrator",
    "msfadmin",
    "user",
]

tier1_passwords = [
    "msfadmin",
    "password",
    "123456",
    "12345678",
    "admin",
]

# Tier 2: Contextual (Metasploitable específico)
tier2_users = [
    "postgres",
    "mysql",
    "ftp",
    "mail",
    "www-data",
    "postgres",
    "test",
    "guest",
]

tier2_passwords = [
    "root",
    "postgres",
    "mysql",
    "test",
    "password123",
    "admin123",
    "1234",
    "pass",
]

print(f"Tier 1 (Padrão): {len(tier1_users)} users × {len(tier1_passwords)} passwords = {len(tier1_users) * len(tier1_passwords)} combinações")
print(f"Tier 2 (Contextual): {len(tier2_users)} users × {len(tier2_passwords)} passwords = {len(tier2_users) * len(tier2_passwords)} combinações")
print(f"\nTotal: {(len(tier1_users) + len(tier2_users)) * (len(tier1_passwords) + len(tier2_passwords))} combinações possíveis")

# Salvar wordlists
wordlists_dir = Path("../wordlists")

with open(wordlists_dir / "ftp_users.txt", "w") as f:
    f.write("\n".join(tier1_users + tier2_users))

with open(wordlists_dir / "ftp_passwords.txt", "w") as f:
    f.write("\n".join(tier1_passwords + tier2_passwords))

print("\n✅ Wordlists salvas em ../wordlists/")
```

## 4. Estratégia de Ataque

```python
print("🎯 PLANO DE ATAQUE\n")

attack_plan = {
    "Fase 1: FTP": {
        "porta": 21,
        "ferramenta": "Medusa",
        "users": len(tier1_users + tier2_users),
        "passwords": len(tier1_passwords + tier2_passwords),
        "threads": 4,
        "tempo_estimado": "5-15 minutos",
        "probabilidade_sucesso": "ALTA (padrões conhecidos)"
    },
    "Fase 2: SMB": {
        "porta": 445,
        "ferramenta": "Medusa",
        "users": len(tier1_users + tier2_users),
        "passwords": len(tier1_passwords + tier2_passwords),
        "threads": 4,
        "tempo_estimado": "10-20 minutos",
        "probabilidade_sucesso": "ALTA (root:root)"
    },
    "Fase 3: Web (DVWA)": {
        "porta": 80,
        "ferramenta": "Burp Suite",
        "usuarios_alvo": ["admin", "user"],
        "passwords": len(tier1_passwords + tier2_passwords),
        "tempo_estimado": "3-5 minutos",
        "probabilidade_sucesso": "ALTA (DVWA por design)"
    }
}

for fase, detalhes in attack_plan.items():
    print(f"\n{fase}")
    for chave, valor in detalhes.items():
        print(f"  {chave}: {valor}")
```

## 5. Estimativa de Tempo

```python
print("⏱️ ESTIMATIVA DE TEMPO DE ATAQUE\n")

# Cálculos
ftp_combinations = len(tier1_users + tier2_users) * len(tier1_passwords + tier2_passwords)
smb_combinations = len(tier1_users + tier2_users) * len(tier1_passwords + tier2_passwords)
web_combinations = 2 * len(tier1_passwords + tier2_passwords)  # 2 users típicos

ms_per_attempt = 500  # 500ms por tentativa (tempo de rede + processamento)

ftp_time = (ftp_combinations * ms_per_attempt) / 1000 / 60  # em minutos
smb_time = (smb_combinations * ms_per_attempt) / 1000 / 60
web_time = (web_combinations * ms_per_attempt) / 1000 / 60

print(f"FTP Brute Force:")
print(f"  Combinações: {ftp_combinations}")
print(f"  Tempo estimado: {ftp_time:.1f} minutos")
print(f"  Com 4 threads paralelos: {ftp_time/4:.1f} minutos\n")

print(f"SMB Brute Force:")
print(f"  Combinações: {smb_combinations}")
print(f"  Tempo estimado: {smb_time:.1f} minutos")
print(f"  Com 4 threads paralelos: {smb_time/4:.1f} minutos\n")

print(f"Web Brute Force (DVWA):")
print(f"  Combinações: {web_combinations}")
print(f"  Tempo estimado: {web_time:.1f} minutos\n")

print(f"TOTAL (sequencial): {ftp_time + smb_time + web_time:.1f} minutos ({(ftp_time + smb_time + web_time)/60:.2f} horas)")
print(f"TOTAL (otimizado com threads): ~{((ftp_time + smb_time + web_time)/4):.1f} minutos")
```
