### 🎯 Objetivo

Compreender como ransomware real funciona:

1. **Infecção** - Usuário executa arquivo malicioso
2. **Enumeração** - Identifica arquivos valiosos
3. **Criptografia** - Criptografa com chave pública
4. **Mensagem de Resgate** - Exibe nota de extorsão
5. **Impacto** - Dados inacessíveis sem descrição

---

### 📚 Conceitos

**Ransomware:**
- Malware que criptografa arquivos do usuário
- Exige pagamento em Bitcoin para descriptografar
- Exemplos reais: WannaCry, Ryuk, LockBit
- Impacto: $20+ bilhões em perdas anuais

**Criptografia Simétrica vs Assimétrica:**
- Simétrica: Mesma chave para encriptar e decriptar
- Assimétrica: Chave pública para encriptar, privada para decriptar
- Ransomware usa assimétrica para manter chave segura

## 1. Setup Inicial

```python
import sys
import os
import json
from pathlib import Path
from datetime import datetime
import logging

# Adicionar scripts ao path
scripts_dir = Path("../scripts/python")
sys.path.insert(0, str(scripts_dir.absolute()))

# Importar simulação de ransomware
from malware.ransomware import RansomwareSimulation

# Setup de logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

# Diretórios
RESULTS_DIR = Path("../results")
TEST_FILES_DIR = RESULTS_DIR / "test_files"
TEST_FILES_DIR.mkdir(exist_ok=True, parents=True)

print(f"✅ Setup concluído")
print(f"   Diretório de teste: {TEST_FILES_DIR}")
print(f"   Timestamp: {datetime.now().isoformat()}")
```

## 2. Criar Arquivos de Teste

Criamos arquivos de teste para serem criptografados. Esses dados são fictícios e isolados.

```python
print("📝 Criando arquivos de teste...\n")

# Inicializar simulação
ransomware = RansomwareSimulation(test_dir=TEST_FILES_DIR)

# Criar arquivos de teste
test_files = ransomware.create_test_files(count=5)

print(f"\n✅ {len(test_files)} arquivos de teste criados:")
for file in test_files:
    file_path = Path(file)
    file_size = file_path.stat().st_size
    print(f"  • {file_path.name} ({file_size} bytes)")

print(f"\n⚠️ Estes arquivos serão criptografados no próximo passo")
```

## 3. Visualizar Conteúdo Original

```python
print("📄 Conteúdo Original dos Arquivos:\n")

for file_path in TEST_FILES_DIR.glob("documento_*.txt"):
    print(f"\n{'='*60}")
    print(f"Arquivo: {file_path.name}")
    print(f"{'='*60}")
    
    with open(file_path, "r") as f:
        content = f.read()
        # Mostrar apenas primeiros 200 caracteres
        display = content[:200] if len(content) > 200 else content
        print(display)
        if len(content) > 200:
            print(f"\n... ({len(content)} bytes total)")
```

## 4. Executar Criptografia (Simulação de Infecção)

⚠️ **CUIDADO**: A partir deste ponto, os arquivos serão criptografados e se tornarão ilegíveis.

```python
print("\n" + "#"*60)
print("🔒 INICIANDO CRIPTOGRAFIA DE TESTE")
print("#"*60)

print("\nEsta simulação reproduz o que um ransomware real faz:")
print("  1. Identifica arquivos valiosos")
print("  2. Criptografa com chave pública")
print("  3. Torna dados inacessíveis")
print("  4. Exibe mensagem de resgate")

print("\n⏳ Processando... Este pode levar alguns segundos\n")

try:
    encryption_result = ransomware.encrypt_files()
    
    print("\n✅ CRIPTOGRAFIA CONCLUÍDA!")
    print(json.dumps(encryption_result, indent=2))
    
except Exception as e:
    print(f"❌ Erro durante criptografia: {e}")
```

## 5. Visualizar Conteúdo Criptografado

Os arquivos agora estão ilegíveis - este é o impacto de um ransomware real.

```python
print("🔐 Conteúdo CRIPTOGRAFADO dos Arquivos:\n")

for file_path in list(TEST_FILES_DIR.glob("documento_*.txt"))[:1]:  # Mostrar apenas um
    print(f"{'='*60}")
    print(f"Arquivo: {file_path.name}")
    print(f"{'='*60}")
    
    # Ler como bytes
    with open(file_path, "rb") as f:
        content = f.read()[:100]  # Primeiros 100 bytes
        print(f"Hex: {content.hex()[:100]}...")
        print(f"\n⚠️ Conteúdo ILEGÍVEL sem chave de descriptografia!")
        print(f"\nTamanho do arquivo: {file_path.stat().st_size} bytes")
        print(f"Status: CRIPTOGRAFADO ❌")
```

## 6. Mensagem de Resgate

Típica de ransomware real - extorsão do usuário.

```python
# Mostrar mensagem de resgate gerada
ransom_file = TEST_FILES_DIR / "LEIA-ME!!!.txt"

if ransom_file.exists():
    print("\n" + "#"*60)
    print("📋 MENSAGEM DE RESGATE ENCONTRADA")
    print("#"*60 + "\n")
    
    with open(ransom_file, "r") as f:
        print(f.read())
else:
    print("⚠️ Mensagem de resgate não encontrada")
```
